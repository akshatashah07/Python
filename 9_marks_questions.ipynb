```python
# vowels and consonants counter
string = input("Enter String : ")
vowels = "AEIOUaeiou"
def count_vc(s):
    v_count = 0
    c_count = 0 
    for i in s:
        if i.isalpha():
            if i in vowels :
                v_count += 1
            else:
                c_count += 1
    return v_count,c_count
x=count_vc(string)
print(x)
print("Vowels : ",x[0])
print("Consonants : ",x[1])
```

```python
# reverse the whole sentence

string = input("Enter String : ")
def reverse_words(s) :
    words = s.split()
    reversed_list = words[::-1]
    return " ".join(reversed_list)
print(reverse_words(string))
```

```python
# masking of vowels
string = input("Enter String : ")
def mask_vowels(s) :
    vowels = "AEIOUaeiou"
    masked_vowels = ""
    for char in s:
        if char in vowels :
            masked_vowels += "*"
        else :
            masked_vowels += char
    return masked_vowels
mask = mask_vowels(string)
print(mask)
```

```python
# code generation
string = input("Enter String : ")
def code_generate(s):
    
    # 1. acronym
    acronym = ""
    words = s.split()
    for word in words:
        if word:
            acronym += word[0].upper()
            
    # 2. digit sum
    sum = 0
    for i in s:
        if i.isdigit():
            sum += int(i)
            
    # 3. vowel parity suffix
    v_count = 0
    vowels = "AEIOUaeiou"
    for i in s :
        if i.isalpha():
            if i in vowels:
                v_count += 1
        if v_count%2 == 0:
            parity = "-EVEN"
        else :
            parity = "-ODD"

    #combine
    code = acronym + str(sum) + parity
    return code

code = code_generate(string)
print(code)
```

```python
def calculate_pizza_price(size, toppings, cheese):
    cost = 0
    # Calculate pizza base price based on size
    if size == 'small':
        cost += 50
    elif size == 'medium':
        cost += 100
    else:
        cost += 200
    
    # Calculate topping prices
    topping_prices_20 = ['corn', 'tomato', 'onion', 'capsicum']
    topping_prices_50 = ['mushroom', 'olives', 'broccoli']
 
    for topping in toppings:
         cost += 20 if topping in topping_prices_20 else 50
    #Calculate cheese price
    cost += 50 * len(cheese)
    return cost

def order_pizza():
    pizzas = []
    number = int(input("How many pizzas you want to order: "))
    for i in range(number):
        toppings = []
        cheese = []
        print('Customize Pizza', i + 1)
        print('Select Pizza Size: small, medium or large')
        size = input('Select size: ')
        
        print('''Select Toppings: 
                20 for 'corn', 'tomato', 'onion', 'capsicum'
                50 for 'mushroom', 'olives', 'broccoli' ''')
        toppings_count = int(input('How many toppings: '))
    
        for _ in range(toppings_count):
             toppings.append(input('Enter toppings: '))
        
        print('Select Cheese: mozarrella, feta, cheddar')
        cheese_count = int(input('How many cheese: '))

        for _ in range(cheese_count):
             cheese.append(input('Enter cheese: '))
        print()
        pizza_cost = calculate_pizza_price(size, toppings, cheese)
        pizzas.append((size, toppings, cheese, pizza_cost))
    return pizzas

def print_bill(pizzas):
     total = 0
     count = 1
     for pizza in pizzas:
         print('Pizza', count)
         print("Size:", pizza[0])
         print("Toppings:", pizza[1])
         print("Cheese:", pizza[2])
         print("Pizza cost:", pizza[3])
         total += pizza[3]
         count += 1
     print('Total bill amount:', total)
# Main program
pizza_order = order_pizza()
print_bill(pizza_order)
```

```python
# product inventory
inventory = {
    "Laptop" : {"price" : 55000, "quantity" : 8, "category" : "Electronics"},
    "Mobile" : {"price" : 25000, "quantity" : 15, "category" : "Electronics"},
    "Chair" : {"price" : 3500, "quantity" : 20, "category" : "Furniture"},
    "Table" : {"price" : 7000, "quantity" : 10, "category" : "Furniture"},
}
```

```python
# 1.) and 2.) Total inventory value of each product -> {product : total_value}
total_value = {item : d["price"] * d["quantity"] for item, d in inventory.items()}
print("1-2) Total value per product : ",total_value)
```

```python
#3.) Products whose total value exceeds Rs 2,00,000
high_value = {item: v for item, v in total_value.items() if v > 200000}
print("3.) Value > 2,00,000 : ",high_value)
```

```python
# 4.) Unique categories using a set
categories = {d["category"] for d in inventory.values()}
print("4.) Unique Categories : ", categories)
```

```python
# 5.) Product with maximum inventory value using lambda
max_item = max(total_value.items(), key = lambda kv:kv[1])
print("5.) Max inventory value : ",max_item[0], "-> Rs", f"{max_item[1]:,}")
```

```python
# 6.) Sort products by inventory value, descending
sorted_items = sorted(total_value.items(),key = lambda kv: kv[1], reverse=True)
print("6.) Sorted (descending) : ")
for name, val in sorted_items:
    print(f" {name:<8} Rs {val:>9,}")
```

```python
# 7.) Extra : dict comprehension -> category-wise total value
category_totals = {c:sum(v for p, v in total_value.items() if inventory[p]["category"] == c)
                  for c in categories}
print("7.) Category-wise totals : ", category_totals)
```

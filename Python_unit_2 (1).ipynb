**Break** - It will break the statement 

**Continue** - It will skip the particular statement

**Pass** - It will do nothing generally 

```python
for i in range(10): #break statement
    if i==7:
        print("Processing Stopped")
        break
    print(i)
```

```python
for i in range(10): #continue statement
    if i==7:
        print("Processing Stopped")
        continue
    print(i)
```

**Write a program to perform arithmetic operations according to the given sign**

```python
a = int(input("Enter number A : "))
b = int(input("Enter number B : "))
choice = input("Enter choice of operation : ")
if choice == '+':
    print("Sum : ", a+b)
elif choice == '-' :
    print("Subtraction : ",a-b)
elif choice == '*' :
    print("Multiplication : ",a*b)
elif choice == '/' :
    print("Divison : ",a/b)
elif choice == '//' :
    print("Floor Division : ",a//b)
elif choice == '%' :
    print("Modulo : ",a%b)
elif choice == '**' :
    print("Exponention : ",a**b)
else :
    print("Write appropriate choice...!!!")
```

**Write a program to calculate electricity bill accept the number of number of units from user**

```python
units = int(input("Enter number of units : "))
bill = 0
if units <= 100 :
    bill = 0
elif units > 100 and units <= 200:
    bill = (units - 100)*5
elif units > 200 :
    bill = 500 + (units-200)*10
print("Total bill : ",bill)
```

**Write a program to check wheather a number entered by user is 3 digit or not if the number is 3 digit then display the middle digit**

```python
n = int(input("Enter three digit number : "))
x = str(n)
count = 0
for i in x :
    count += 1
if count == 3 :
    print("Middle digit is : ",(n//10)%10) #((n%100)//10)
else:
    print("Number is not 3 digit...!!!")
```

**Wrirte a program to find sum of n numbers**

```python
n = int(input("Enter number : "))
sum = 0
for i in range(1, n+1) :
    sum = sum + i
print("Sum : ",sum)
print("Average : ", sum/n)
```

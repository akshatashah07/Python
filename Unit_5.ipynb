```python
l = [1,2,3,4,5]
l[1] = 10
print(l)
```

```python
l = []
print(type(l))
```

```python
l = [10]
print(type(l))
```

```python
l = list(range(1,11))
print(l)
s = "aryan"
print(list(s))
```

```python
#eval function(Dynamic Input)
l = eval(input("Enter data : "))
print(l)
print(type(l))
```

```python
#Accessing  the elements of list
# 1. By using index
l = [1,2,3,4,5,6]
print(l[1])
print(l[-2])
#print(l[7])
```

```python
# 2. By using slicing operator
l = list(range(1,11))
print(l)
print(l[2:7:2])
print(l[4::2])
print(l[3:7])
print(l[8:2:-2])
print(l[4:100])
print(l[::-1])
print(l[-8:-2:-2])
print(l[-8:-2:2])
```

```python
# Mathematical Operators of list

l1 = [1,2,3]
l2 = [4,5,6]
print(l1+l2)
print(l1*3)
```

```python
c = l1 + 4
print(c)
```

```python
c = l1 + [4]
print(c)
```

```python
# Comparison operator for list
x = ["Dog", "Cat", "Rat"]
y = ["Dog", "Cat", "Rat"]
z = ["DOG", "CAT", "RAT"]
print(x==y)
print(x==z)
print(x!=z)
```

```python
x = [50,20,30]
y = [40,50,60,100,120]
print(x>y)
print(x>=y)
print(x<y)
print(x<=y)
```

```python
x = ['Dog','Cat','Rat']
y = ['Rat','Cat','Cat']
print(x>y)
print(x>=y)
print(x<y)
print(x<=y)
```

```python
x = ['dog','Cat','Rat']
y = ['Rat','Cat','Cat']
print(x>y)
print(x>=y)
print(x<y)
print(x<=y)
```

```python
# Membership operator for list
l = [1,2,3,4,5]
print(6 in l)
print(5 in l)
print(7 not in l)
```

```python
# Aliasing and cloning of list
s = [10,20,30,40,50]
x = s
x[3] = 100
print(s)
print(id(s))
print(id(x))
```

```python
s = [10,20,30,40,50]
x = s[:]
x[3] = 100
print(s)
print(x)
print(id(s))
print(id(x))
```

```python
s = [10,20,30,40,50]
x = s.copy()
x[3] = 100
print(s)
print(x)
print(id(s))
print(id(x))
```

### Important Functions of list

```python
#len()
l = [1,2,3,4,5]
print(len(l))
```

```python
#count
l = [1,1,1,2,3,3,4,4,4,4,5]
print(l.count(1))
print(l.count(4))
print(l.count(6))
```

```python
# index
l = [1,1,1,2,3,3,4,4,4,4,5]
print(l.index(3))
#print(l.index(6))
print(l.index(3,5,))
if 6 in l:
    print(l.index(6))
```

```python
# append()
l = [1,2,5,6,8]
l.append(4)
print(l)
l.append("abcd")
print(l)
l.append([1,2,3])
print(l)
```

```python
# insert()
l = [1,2,5,6,8]
l.insert(2,10)
#print(l.insert(2,10))
print(l)
l.insert(2,9)
print(l)
l.insert(-10,11)
print(l)
l.insert(50,12)
print(l)
l.insert(-1,13)
print(l)
```

```python
l1 = [1,2,3]
l2 = [4,5,6]

l1.extend(l2)
print(l1)
print(l2)
print("------------")
l2.extend(l1)
print(l1)
print(l2)
```

```python
l1 = [1,2,3]
l1.extend("LJU")
print(l1)
```

```python
# remove()
n = [1,2,1,3,1,4,1,5]
n.remove(1)
print(n)
```

```python
for i in n:
    if i==1:
        n.remove(1)
print(n)
```

```python
l = [1,1,1,1,3,4,6,9,1,3,1,4,1,5,2]
while 1 in l :
    l.remove(1)
print(l)
```

```python
#pop()
l = [10,20,30,40,50]
#print(l.pop())
l.pop()
print(l)
```

```python
l = [12,234,6,7678,345]
l.pop(0)
print(l)
```

```python
# clear()
l = [10,20,30,40,50]
l.clear()
print(l)
```

```python
# reverse()
l = [10,20,30,40,50]
l.reverse()
print(l)
```

```python
# sort()
l = [10,45,35,34,23]
l.sort()
print(l)
```

```python
l = [10,45,35,34,23]
l.sort(reverse = True)
print(l)
```

```python
l = ["A","F","C","N","a","b"]
l.sort()
print(l)
```

```python
l = ["A","F","C","N","a","b"]
l.sort(reverse = True)
print(l)
```

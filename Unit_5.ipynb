## List Comprehension

#### syntax : list = [expression for item in list if condition]

```python
s = [x*x for x in range(1,11)]
print(s)
```

```python
l = list(range(1,11,2))
s = [x*x for x in l]
print(s)
```

```python
s = [x*x for x in range(1,11) if x%2==0]
print(s)
```

```python
l = [22,11,35,46,78,98,65,43]
d = [x for x in l if x%2==0]
print(d)
```

```python
result = [x*y
         for x in range(1,5)
         for y in range(1,4)
         if (x+y) % 2 == 0]
print(result)
```

**Write a program to perform a circular shift on a list to the right direction (shift is to be entered by a user)**

```python
l = eval(input("Enter elemets : "))
shift = int(input("Enter numbers of shift : "))
n_shift = shift % len(l)

for i in range(0,n_shift) :
    elem = l.pop()
    l.insert(0,elem)

print(l)
```

```python
l = eval(input("Enter elemets : "))
shift = int(input("Enter numbers of shift : "))
n_shift = shift % len(l)
x = l[-shift:]+l[:-shift]
print(x)
```

**special element -> removal of element makes list balanced**

**list will be balanced if sum of even index elements is equal to sum of odd index elements**

**given a list l of size l you need to count the number of special element in the given list an element is special if the removal of that element makes the list balanced the list will be balanced if sum of even index elements is equal to sum of odd index elements also display the updated list after removal of special element**

```python
l = [1,2,3,4,5]
print(sum(l))
```

```python
l = [5,5,2,5,8]
print("Original List : ",l)
print()
count = 0
for i in range(len(l)):
    c = l.copy()
    c.pop(i)
    sum1 = sum(c[0::2])
    sum2 = sum(c[1::2])
    if sum1 == sum2:
        print("Index to be removed : ",i)
        count += 1
        print("List after removal of index",i,":",l[:i]+l[i+1:])
        print()
print("Total special elements :",count)
```

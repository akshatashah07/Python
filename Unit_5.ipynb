```python
s = "dog the quick brown fox jumps over the dog"
words = s.split()
d = {}
for word in words :
    key = word[0]
    if key in d:
        d[key].append(word)
    else:
        d[key] = [word]
print(d)
```

**Write a program to replace the words in a given string from the lookup(key and value replace with value) dictionary**

```python
s = "Don't wait for your feelings to change to take action. Take the action as your feelings will change."
d = {"action" : "lazy", "feelings" : "rude"}
new = []
x = s.split()
for i in x : #for i in s.split()
    if i in d :
        new.append(d[i])
    else :
        new.append(i)
x = " ".join(new)
print(x)
```

## Set

- Mutable and unordered
- if we want to represent a group of unique values as a single entity then we should go for set.
- duplicates are not allowed
- insertion order is not preserved but we can sort the elements.
- indexing and slicing are not allowed.
- heterogenous objects are allowed.
- we can represent set elements within curly brackets with comma separator.
- we can apply mathematical operations like union, intersection, difference, symmetric difference etc... on set objects.

```python
# Creation of set

s = {1,1,2,3,4,4,4,4,5,6}
print(s)
print(type(s))
```

### Mathematical Operations on set

```python
# 1.) union()

x = {10,20,30,40}
y = {30,40,50,60}
print(x.union(y))
print(x|y)
print()
print(x)
print(y)
```

```python
# 2.) intersection

x = {10,20,30,40}
y = {30,40,50,60}
print(x.intersection(y))
print(x&y)
```

```python
# 3.) difference

x = {10,20,30,40}
y = {30,40,50,60}
print(x.difference(y))
print(y.difference(x))
print(x-y)
print(y-x)
```

```python
# 4.) symmetric_difference()

x = {10,20,30,40}
y = {30,40,50,60}
print(x.symmetric_difference(y))
print(x^y)
```

```python
# 5.) issubset()
x = {'a','b','c'}
y = {'e','f','a','b','c'}
print(x.issubset(y))
print(y.issubset(x))
```

```python
# 5.) issuperset()
x = {'a','b','c'}
y = {'e','f','a','b','c'}
print(x.issuperset(y))
print(y.issuperset(x))
```

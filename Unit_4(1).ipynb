**Enumerate() Function :**

Syntax : enumerate(iterable, start=0)

```python
l = ["Arman", "Aryan", "Dhairya"]
obj1 = list(enumerate(l))
print(obj1)
```

```python
for i in obj1 :
    print(i)
```

```python
l = ["Arman", "Aryan", "Dhairya"]
obj1 = list(enumerate(l,5))
print(obj1)
```

```python
for i,j in obj1 :
    print(i,j)
```

```python
s = "adjhasgdjhsvfsdjvf"
obj2 = list(enumerate(s))
print(obj2)
```

```python
for i,j in obj2:
    print(i,j)
```

```python
s = "bgfgbheruighhgefh"
obj2 = enumerate(s)
count = 0
pos = 0

for i in obj2:
    if i[1]=="g" :
        count += 1
        pos = i[0]
        print("Position : ",pos)

print("Total Count : ",count)
```

**Write a program to print sum of even numbers and sum odd numbers from the given elements in a tuple**

```python
t = (10,12,15,19,22,51,65,55,54,88)
even = 0
odd = 0
for i in t :
    if i%2 == 0 :
        even += i
    else :
        odd += i
print("Even : ",even)
print("Odd : ",odd)
```

# Different categories of UDF #



**1.) Function with no parameter and no return value**

```python
def printline():
    s = input("Enter name : ")
    print(s)
printline()
```

**2.) Function with parameter and no return value**

```python
def printline(s):
    print(s)
s = input("Enter Name : ")
printline(s)
```

**2.) Function with parameter and with return value**

```python
def printline(s):
    return s
s = input("Enter Name : ")
t = printline(s)
print(t) # print(printline(s))
```

**2.) Function with no parameter and with return value**

```python
def printline():
    s = input("Enter Name : ")
    return s
print(printline()) # t = printline()
                    # print(t)
```

### Types of Argument ###

```python
def f1(a,b):
    print(a+b)
f1(10,20)
# a,b -> Formal Argument
# 10,20 -> Actual Argument
```

**1.) Positional Argument**

```python
def sub(a,b):
    print(a-b)
sub(100,200)
sub(200,100)
```

**2.) Keyword Argument**

```python
def wish(name, msg):
    print("Hello", name, msg)
wish(name = "Arman", msg = "Good Morning")
wish(msg = "Good Morning", name = "Aryan")
```

```python
wish(msg = "Good Morning", name = "Aryan")
```

```python
wish(name="Arman","Good Morning")
```

**3.) Default Argument**

```python
def wish(name = "Guest"):
    print("Hello", name)
wish("Arman")
wish()
```

```python
def wish(name = "Guest", msg = "Good Morning"):
    print("Hello", name, msg)
wish("Arman")
wish()
```

```python
def wish(name, msg = "Good Morning"):
    print("Hello", name, msg)
wish("Arman")
wish()
```

```python
def wish(name = "Guest", msg):
    print("Hello", name, msg)
wish("Arman")
```

**4.) Variable Length Argument**

```python
def sum(*n): # *args
    total = 0
    for n1 in n:
        total = total + n1
    print("Total : ", total)
sum(10)
sum(10,20,30)
sum(40,60,70,80,90)
```

```python
def f1(n1,*s):
    print(n1)
    for s1 in s:
        print(s1)
f1(10)
f1(10,20,30,40)
```

```python
def f1(*s,n1):
    for s1 in s:
        print(s1)
    print(n1)
f1(n1=10)
f1(10,20,30,n1=40)
f1(10,"A","B","D",50)
```

```python
def display(**n):
    print(n)
    for k,v in n.items():
        print(k,"=",v)
display(m1=100,m2=200,m3=300)
display(rno=1, name="Aryan", marks=80, subject = "python")
```

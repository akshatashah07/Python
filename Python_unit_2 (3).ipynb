**Fibonacci Series**

**Reverse Number**


```python
n = int(input("Enter Number : "))
rev = 0
while n!=0:
    digit = n%10
    rev = rev*10+digit
    n = n//10
    print(rev)
```

**Palindrome number**

```python
n = int(input("Enter Number : "))
temp = n
rev = 0
while n!=0:
    digit = n%10
    rev = rev*10+digit
    n = n//10
    print(rev)
if rev == temp :
    print("The number is palindrome number.")
else :
    print("The number is not palindrome number.")
```

**Armstrong Number**


```python
n = int(input("Enter Number: "))

temp = n
count = 0

# Count number of digits
while n != 0:
    count = count + 1
    n = n // 10

# Reset n
n = temp
rev = 0

# Calculate Armstrong sum
while n != 0:
    digit = n % 10
    rev = rev + (digit ** count)
    n = n // 10

print("Armstrong sum:", rev)

if rev == temp:
    print("The number is Armstrong number.")
else:
    print("The number is not Armstrong number.")
```

**Disarium Number**

```python
n = int(input("Enter Number: "))

temp = n
count = 0

# Count number of digits
while n != 0:
    count = count + 1
    n = n // 10
print(count)
# Reset n
n = temp
rev = 0

# Calculate Disarium sum
for i in range(count,0,-1):
    digit = n % 10
    rev = rev + (digit ** i)
    n = n // 10
print(rev)
# print("Armstrong sum:", rev)

if rev == temp:
     print("The number is Dissarium number.")
else:
     print("The number is not Dissarium number.")
```

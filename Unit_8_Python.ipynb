### Exception Handling 

Error in python can be of two types
1) Compile Time (syntax errors, indentation errors)
2) Run Time

- Errors are problems in a program due to which the program will stop execution.
- On the other hand, exceptions are raised when some internal events occur which change the normal flow of the program.

```python
def div(a,b):
    return a/b
print(div(5,0))
```

### try except else finally
- The try block lets you test a block of code for errors.
- The except block lets you handle the error.
- The else block lets you execute code when there is no error.
- The finally block lets you execute the code regardless of the result of the try and except blocks.

```python
def div(a,b):
    return a/b
try :
    print(div(5,3))
except :
    print("Exception Occured")
else :
    print("No Exeception")
finally :
    print("This always prints")
```

```python
try :
    m = 5
    f = open("abc.txt","w+")
    print(f.read())
    print(m)
    print(5/2)
    l = [1,2,3]
    print(l[100])
except FileNotFoundError :
    print("Text File is not Found")
except NameError :
    print("Variable is not defined")
except ZeroDivisionError :
    print("Can't Divide by Zero")
except IndexError :
    print("Index out of range")
```

```python
try :
    m = 5
    f = open("abcd.txt","r")
    print(f.read())
    print(m)
    print(5/0)
    l = [1,2,3]
    print(l[100])
except Exception as e:
    print(e)
```

```python
try :
    f = open("abc.txt","r")
    print(f.read())
    print(m)
    print(5/0)
    l = [1,2,3]
    print(l[100])
except Exception as d :
    print(d)
except (FileNotFoundError, ZeroDivisionError) as e:
    print(e)
```

## Raise an Exception

```python
def add(a,b) :
    if type(a) == str:
        raise TypeError("a should be integer")
    if type(b) == str:
        raise TypeError("b should be integer")
    return a+b
print(add(10,20))
```

```python
def add(a,b) :
    if type(a) == str:
        raise IndexError("a should be integer")
    if type(b) == str:
        raise IndexError("b should be integer")
    return a+b
print(add("10",20))
```

```python
class akshata(Exception):
    def __init__(self,arg):
        self.msg = arg
```

```python
def add(a,b) :
    if type(a) == str:
        raise akshata("Personalized Error Message")
    if type(b) == str:
        raise TypeError("b should be integer")
    return a+b
print(add("10","20"))
```

```python
def checkbalance():
    money = int(input("Enter Balance : "))
    withdraw = int(input("Enter Amount to Withdraw : "))
    try :
        if money>withdraw :
            balance = money - withdraw
            if balance <= 2000 :
                raise akshata("Low Balance Alert !!!")
            print("Total balance after withdrawal : ",balance)
        else :
            print("Unsufficient Balance")
            print("Total Balance : ",money)
    except akshata as e :
        print(e)
checkbalance()
```

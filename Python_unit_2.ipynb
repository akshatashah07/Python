**Simple if**

if condition:

    statement 1
    statement 2

```python
name = input("Enter Name : ")
if name == "Arman":
    print("Hello Arman !!! ")
print("How are you ??")
```

```python
# Example of if-else
name = input("Enter Name : ")
if name == "Arman":
    print("Hello Arman !!! ")
else :
    print("Hello Guest !!! ")
print("How are you ?? ")
```

```python
# Example of if-elif-else
n1 = int(input("Enter n1 : "))
n2 = int(input("Enter n2 : "))
n3 = int(input("Enter n3 : "))
if n1>n2 and n1>n3 :
    print("Biggest number is n1 : ", n1)
elif n2>n3 :
    print("Biggest number is n2 : ", n2)
else :
    print("Biggest number is n3 : ", n3)
```

```python
# Example of nested if
x = int(input("Enter Number : "))
if x>10 :
    print("Above 10")
    if x>20 :
        print("Also Above 20")
    else :
        print("But not above 20")
else :
    print("Below 10")
```

**Iterative Statements**

##### 1.) For Loop
Syntax : for x in sequence :
            
            body of loop

```python
x = "Arman"
for i in x:
    print(i)
```

```python
for i in range (0, 50, 10):
    print(i)
```

```python
for i in range (1, 11) :
    if i!=3 and i!=6:
        print(i)
```

```python
# While loop
x = 1
while x<11 :
    print(x)
    x+=1
```

# Different categories of UDF #



**1.) Function with no parameter and no return value**

```python
def printline():
    s = input("Enter name : ")
    print(s)
printline()
```

**2.) Function with parameter and no return value**

```python
def printline(s):
    print(s)
s = input("Enter Name : ")
printline(s)
```

**2.) Function with parameter and with return value**

```python
def printline(s):
    return s
s = input("Enter Name : ")
t = printline(s)
print(t) # print(printline(s))
```

**2.) Function with no parameter and with return value**

```python
def printline():
    s = input("Enter Name : ")
    return s
print(printline()) # t = printline()
                    # print(t)
```

### Types of Argument ###

```python
def f1(a,b):
    print(a+b)
f1(10,20)
# a,b -> Formal Argument
# 10,20 -> Actual Argument
```

**1.) Positional Argument**

```python
def sub(a,b):
    print(a-b)
sub(100,200)
sub(200,100)
```

**2.) Keyword Argument**

```python
def wish(name, msg):
    print("Hello", name, msg)
wish(name = "Arman", msg = "Good Morning")
wish(msg = "Good Morning", name = "Aryan")
```

```python
wish(msg = "Good Morning", name = "Aryan")
```

```python
wish(name="Arman","Good Morning")
```

**3.) Default Argument**

```python
def wish(name = "Guest"):
    print("Hello", name)
wish("Arman")
wish()
```

```python
def wish(name = "Guest", msg = "Good Morning"):
    print("Hello", name, msg)
wish("Arman")
wish()
```

```python
def wish(name, msg = "Good Morning"):
    print("Hello", name, msg)
wish("Arman")
wish()
```

```python
def wish(name = "Guest", msg):
    print("Hello", name, msg)
wish("Arman")
```

**4.) Variable Length Argument**

```python
def sum(*n): # *args
    total = 0
    for n1 in n:
        total = total + n1
    print("Total : ", total)
sum(10)
sum(10,20,30)
sum(40,60,70,80,90)
```

```python
def f1(n1,*s):
    print(n1)
    for s1 in s:
        print(s1)
f1(10)
f1(10,20,30,40)
```

```python
def f1(*s,n1):
    for s1 in s:
        print(s1)
    print(n1)
f1(n1=10)
f1(10,20,30,n1=40)
f1(10,"A","B","D",50)
```

```python
def display(**n):
    print(n)
    for k,v in n.items():
        print(k,"=",v)
display(m1=100,m2=200,m3=300)
display(rno=1, name="Aryan", marks=80, subject = "python")
```

```python
year = int(input("Enter a year : "))
leap_year = False
if (year%4 == 0) and (year%100 != 0 or year%400 == 0) :
    leap_year = True
month = int(input("Enter a month [1-12] : "))
if month<=12 :
    if month in (1,3,5,7,8,10,12):
        length = 31
    elif month == 2:
        if leap_year :
            length = 29
        else :
            length = 28
    elif month in (4,6,9,11):
        length = 30
    day = int(input("Enter a day [1-31] : "))
    if day>length :
        print("Enter valid day.")
    else :
        if day<length :
            day += 1
            print(f"Next Date : {day}-{month}-{year}")
        else :
            day = 1
            if month == 12:
                month = 1
                year += 1
            else :
                month += 1
            print(f"Next Date : {day}-{month}-{year}")
else :
    print("Enter appropriate month.")
```

```python
# Harshad Number
# 156 -> 1+5+6 = 12 , 156%12 == 0 then harshad number otherwise not
def harshad(n):
    temp = n
    sum = 0
    while n!=0:
        d = n%10
        sum = sum+d
        n = n//10
    if temp % sum == 0:
        print("Harshad Number.")
    else :
        print("Not a harshad number.")
n = int(input("Enter a number : "))
harshad(n)
```

## Creating a Directory
### os.mkdir()
- This method is used to create a directory named path with the specified folder. This method raises **FileExistsError** if the directory to be created is already exists.

## Listing out files and directories in python
### **os.listdir()**
- This method is used to get the list of all files and directories in the specified directory.
- If we don't specify directory then the list of files and directories in the current working directory will be returned

```python
import os
os.listdir("D:\\B9")
```

## Deleting the Directory or Files using Python
- OS module provides different methods for removing directories and files in python
    - **os.remove()**
    - **os.rmdir()**

#### os.remove()
- This method is used to remove or delete a file. This method can not remove or delete a directory. If specified path is directory then OSError will be raised by the method.
#### os.rmdir()
- This method is used to remove or delete a directory. If any file exists under the directory it will raise an error.

```python
import os
os.remove('output.txt')
```

# Modules
- A Python module is a file containing python definitions and statements. A Module can define functions, class and variables. Is makes the code logically organized.

```python
%%writefile mymodule.py
def add(a,b):
    print(a+b)
def sub(a,b):
    print(a-b)
def mul(a,b):
    print(a*b)
def div(a,b):
    print(a/b)
def power(a,b):
    print(a**b)
def floor(a,b):
    print(a//b)
def modulo(a,b):
    print(a%b)
```

```python
%%writefile editstr.py
def editstring(s):
    print("Uppercase : ",s.upper())
    print("Lowercase : ",s.lower())
    print("Capitalize : ",s.capitalize())
    print("Titlecase",s.title())
    print("Swapcase : ",s.swapcase())
    print("isalpha : ",s.isalpha())
    print("isalnum : ",s.isalnum())
    print("isdigit : ",s.isdigit())
```

```python
import editstr
s = input("Enter String : ")
editstr.editstring(s)
```

### Datetime Module
- **date** - represents a calendar date (year, month, day)
- **time** - represent a time of day (hour, minute, second usec) and optional timezone
- **datetime** - combines both date and time information.

```python
from datetime import date
today = date.today()
print(today)
```

```python
year, month, day = today.year, today.month, today.day
print(year, month, day)
```

```python
from datetime import date
d = date(2026,10,2)
print(d)
```

```python
from datetime import date
today = date.today()
print(today.strftime("%d %b %y (%A)"))
print(today.strftime("%d %B %y (%A)"))
print(today.strftime("%d %B %y (%a)"))
```

```python
print(today.strftime("%d/%m/%y"))
print(today.strftime("%d/%m/%Y"))
```

```python
from datetime import time
t = time(12,10,0)
print(t)
```

```python
from datetime import datetime
now = datetime.now()
print(now)
```

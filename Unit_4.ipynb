**Write a program to check wheather the password is valid or not**

**1.) Minimum 8 characters**

**2.)The alphabet must be between [a-z]**

**3.)Atleast one alphabet should be uppercase[A-Z]**

**4.)Atleast one digit between [0-9]**

**5.)Atleast one special symbol from (_ or @ or $)**

```python
s = input("Enter Password : ")
uppercase = False
alphabets = 0
digits = False
if len(s) >= 8 :
    for i in s :
        if i.isalpha() :
            alphabets += 1
            if i.isupper() :
                uppercase = True
            else :
                print("There should be atleast one uppercase letter between [A-Z]...!!!")
        if i.isdigit():
            digits = True
        else :
            print("There should be atleast one digit between [0-9]")
    print("Password is valid")
else :
    print("Incorrect Password !!! Enter valid password")
```

```python
pw = input("Enter Password : ")
l,u,sp,d = 0,0,0,0
if len(pw) >= 8:
    for i in pw :
        if i.islower():
            l += 1
        if i.isupper():
            u += 1
        if i.isdigit():
            d += 1
        if (i=="@" or i=="$" or i=="_") :
            sp += 1
    if l>0 and u>0 and sp>0 and d>0 and (l+u+sp+d) == len(pw) :
        print("Valid Password.")
    else :
        print("Invalid Password")
else :
    print("Enter more than 8 characters")
```

# Tuple

**Tuple is immutable. Tuple is Read only version of list.**

**If data is fixed and never changes then we can go for tuple.**

**Insertion order is preserved.**

**Duplicates are allowed.**

**Heterogenous objects are allowed.**

**We can differentiate duplicate objects by using indexing.**

**Tuple supports both +ve and -ve indexing.**

**Represent tuple elements within paranthesis (()) with comma separator.**

**Paranthesis are optional but recommended to use.**

#### Creation of Tuple

```python
t = ()
print(type(t))
```

```python
t = (10,) # t = (10) it will give int as a type
print(type(t))
```

```python
t = 10,20,30,40
print(t)
```

```python
t = tuple(range(10,20,2))
print(t)
```

### Accessing the elements of Tuple

**1.) By using index**

**2.) By using slicing operation**

```python
t = (10,20,30,40,50,60)
print(t[3])
print(t[-3])
print(t[-5])
print(t[7])
```

```python
t = (10,20,30,40,50,60)
print(t[2:5])
print(t[2:100])
print(t[:100])
print(t[::2])
print(t[5:])
print(t[-5:-1])
print(t[-5:])
print(t[-5:-1:-1])
print(t[5:1:1])
print(t[5:1:-1])
```

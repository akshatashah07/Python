**8.) isspace()**

```python
x = " "
print(x.isspace())
```

```python
count = 0
s = "LJ University LJIET Engineering"
for i in s :
    if i.isspace() :
        count = count + 1
print(count)
```

```python
alphabets = 0
upper = 0
lower = 0
digits = 0
special_symbols = 0
s = "Py@$%3725t@hon"
for i in s :
    if i.isalpha() :
        alphabets = alphabets + 1
        if i.isupper() :
            upper = upper + 1
        if i.islower() :
            lower = lower + 1
    elif i.isdigit() :
        digits = digits + 1
    elif i.isspace() :
        pass
    else :
        special_symbols = special_symbols + 1 
        
print(alphabets)
print(upper)
print(lower)
print(digits)
print(special_symbols)
```
